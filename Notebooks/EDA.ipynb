```python
import plotly.express as px
```

```python




def plot_target_pie(df):
    fig = px.pie(
        df,
        names='is_canceled',
        title='Canceled vs Not Canceled Bookings'
    )
    fig.show()


def plot_hotel_cancellation_percentage(df):
    counts = (
        df.groupby(['hotel', 'is_canceled'])
        .size()
        .reset_index(name='count')
    )

    counts['percentage'] = (
        counts['count'] /
        counts.groupby('hotel')['count'].transform('sum') * 100
    )

    fig = px.bar(
        counts,
        x='hotel',
        y='percentage',
        color='is_canceled',
        barmode='stack',
        text=counts['percentage'].round(1),
        title='Hotel Type vs Cancellation Percentage'
    )

    fig.update_layout(yaxis_title='Percentage (%)')
    fig.show()




def plot_lead_time_hist(df):
    fig = px.histogram(
        df,
        x='lead_time',
        color='is_canceled',
        nbins=50,
        title='Lead Time Distribution by Cancellation'
    )
    fig.show()




def plot_customer_type_pie(df):
    fig = px.pie(
        df,
        names='customer_type',
        title='Customer Type Distribution'
    )
    fig.show()



def plot_special_requests(df):
    fig = px.histogram(
        df,
        x='total_of_special_requests',
        color='is_canceled',
        title='Special Requests vs Cancellation'
    )
    fig.show()



import pandas as pd 
df=pd.read_csv(r"C:\Users\User\Desktop\AI_Projects\Project_09\Data\Raw\hotel_bookings_updated_2024.csv")

plot_target_pie(df)
plot_hotel_cancellation_percentage(df)
plot_lead_time_hist(df)
plot_customer_type_pie(df)
plot_special_requests(df)

```

```python
import plotly.express as px

def plot_missing_values(df):
    missing = df.isnull().sum()

    missing = missing[missing > 0]

    if missing.empty:
        print("No missing values found in the dataset.")
        return

    missing_df = (
        missing
        .reset_index()
        .rename(columns={'index': 'column', 0: 'missing_count'})
    )

    fig = px.bar(
        missing_df,
        x='column',
        y='missing_count',
        title='Missing Values by Column'
    )

    fig.update_layout(
        xaxis_title='Column',
        yaxis_title='Number of Missing Values'
    )

    fig.show()
plot_missing_values(df)
```
